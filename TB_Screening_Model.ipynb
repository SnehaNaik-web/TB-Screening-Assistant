```python
import os
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from concurrent.futures import ThreadPoolExecutor, as_completed

# Step 1: Rebuild the official image URL list
index_url = (
    "https://data.lhncbc.nlm.nih.gov/public/"
    "Tuberculosis-Chest-X-ray-Datasets/"
    "Shenzhen-Hospital-CXR-Set/CXR_png/index.html"
)

response = requests.get(
    index_url,
    headers={"User-Agent": "Mozilla/5.0"},
    timeout=30
)
response.raise_for_status()

soup = BeautifulSoup(response.text, "html.parser")

image_urls = [
    urljoin(index_url, a["href"])
    for a in soup.find_all("a", href=True)
    if a["href"].lower().endswith(".png")
]

print("Image links found:", len(image_urls))

# Step 2: Create the dataset folder
folder = "/content/tb_dataset"
os.makedirs(folder, exist_ok=True)

# Step 3: Download the images
def download_image(url):
    filename = os.path.basename(url)
    filepath = os.path.join(folder, filename)

    if os.path.exists(filepath) and os.path.getsize(filepath) > 0:
        return True

    try:
        r = requests.get(url, timeout=60)
        r.raise_for_status()

        if not r.content.startswith(b"\x89PNG\r\n\x1a\n"):
            return False

        with open(filepath, "wb") as f:
            f.write(r.content)

        return True
    except Exception as e:
        print("Download failed:", filename, str(e))
        return False

completed = 0
failed = 0

with ThreadPoolExecutor(max_workers=4) as executor:
    futures = [executor.submit(download_image, url)
               for url in image_urls]

    for future in as_completed(futures):
        if future.result():
            completed += 1
        else:
            failed += 1

        if (completed + failed) % 50 == 0:
            print(f"Processed {completed + failed}/{len(image_urls)}")

print("\nDownload process finished!")
print("Images downloaded or already present:", completed)
print("Failed downloads:", failed)
print("Images in folder:", len([
    f for f in os.listdir(folder) if f.endswith(".png")
]))
```

```python

import os
import torch
import numpy as np

from PIL import Image
from torch import nn
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
from torchvision.models import DenseNet121_Weights
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score
)

# 1. Check GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# 2. Prepare image paths and labels
folder = "/content/tb_dataset"
image_paths = []
labels = []

for filename in sorted(os.listdir(folder)):
    if not filename.endswith(".png"):
        continue

    if filename.endswith("_0.png"):
        label = 0  # Normal
    elif filename.endswith("_1.png"):
        label = 1  # TB-consistent abnormality
    else:
        continue

    image_paths.append(os.path.join(folder, filename))
    labels.append(label)

print("Total images:", len(image_paths))

# 3. Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    image_paths,
    labels,
    test_size=0.20,
    random_state=42,
    stratify=labels
)

print("Training images:", len(X_train))
print("Testing images:", len(X_test))

# 4. Image preprocessing
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# 5. Create a PyTorch dataset
class ChestXrayDataset(Dataset):
    def __init__(self, paths, labels, transform):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, idx):
        image = Image.open(self.paths[idx]).convert("RGB")
        image = self.transform(image)
        return image, self.labels[idx]

train_dataset = ChestXrayDataset(
    X_train, y_train, train_transform
)
test_dataset = ChestXrayDataset(
    X_test, y_test, test_transform
)

train_loader = DataLoader(
    train_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

test_loader = DataLoader(
    test_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=2,
    pin_memory=torch.cuda.is_available()
)

# 6. Load DenseNet121 pretrained on ImageNet
weights = DenseNet121_Weights.DEFAULT
model = models.densenet121(weights=weights)

# Freeze the pretrained feature extractor
for param in model.features.parameters():
    param.requires_grad = False

# Replace final layer for our two classes
in_features = model.classifier.in_features
model.classifier = nn.Linear(in_features, 2)
model = model.to(device)

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(
    model.classifier.parameters(),
    lr=0.001
)

# 7. Train the classification layer
epochs = 5

for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    for images, targets in train_loader:
        images = images.to(device)
        targets = targets.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * images.size(0)
        predictions = outputs.argmax(dim=1)
        correct += (predictions == targets).sum().item()
        total += targets.size(0)

    epoch_loss = running_loss / total
    epoch_accuracy = correct / total

    print(
        f"Epoch {epoch + 1}/{epochs} | "
        f"Loss: {epoch_loss:.4f} | "
        f"Training accuracy: {epoch_accuracy:.4f}"
    )

# 8. Evaluate on the held-out test set
model.eval()
all_labels = []
all_predictions = []
all_probabilities = []

with torch.no_grad():
    for images, targets in test_loader:
        images = images.to(device)
        outputs = model(images)
        probabilities = torch.softmax(outputs, dim=1)

        all_labels.extend(targets.numpy().tolist())
        all_predictions.extend(
            outputs.argmax(dim=1).cpu().numpy().tolist()
        )
        all_probabilities.extend(
            probabilities[:, 1].cpu().numpy().tolist()
        )

print("\nConfusion matrix:")
print(confusion_matrix(all_labels, all_predictions))

print("\nClassification report:")
print(classification_report(
    all_labels,
    all_predictions,
    labels=[0, 1],
    target_names=["Normal", "TB-consistent abnormality"],
    zero_division=0
))

if len(set(all_labels)) == 2:
    print(
        "ROC-AUC:",
        round(roc_auc_score(all_labels, all_probabilities), 4)
    )

# 9. Save model weights
model_path = "/content/densenet121_tb_screening.pth"
torch.save(model.state_dict(), model_path)
print("\nModel saved to:", model_path)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Keep early feature layers frozen
for param in model.features.parameters():
    param.requires_grad = False

# Unfreeze only the final feature block
for param in model.features.denseblock4.parameters():
    param.requires_grad = True

# Also unfreeze its final normalization layer
for param in model.features.norm5.parameters():
    param.requires_grad = True

# Keep the classifier trainable
for param in model.classifier.parameters():
    param.requires_grad = True

# Use a smaller learning rate for fine-tuning
optimizer = optim.Adam([
    {
        "params": model.features.denseblock4.parameters(),
        "lr": 1e-5
    },
    {
        "params": model.features.norm5.parameters(),
        "lr": 1e-5
    },
    {
        "params": model.classifier.parameters(),
        "lr": 1e-4
    }
])

criterion = nn.CrossEntropyLoss()

print("Fine-tuning setup completed!")

print(
    "Trainable parameters:",
    sum(p.numel() for p in model.parameters() if p.requires_grad)
)

print("Device:", next(model.parameters()).device)
```

```python
from tqdm.auto import tqdm
import torch

# Move model to GPU
model = model.to(device)

# Fine-tuning for 5 epochs
num_epochs = 5

for epoch in range(num_epochs):

    model.train()

    running_loss = 0.0
    correct = 0
    total = 0

    progress_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}")

    for images, labels in progress_bar:

        images = images.to(device)
        labels = labels.to(device)

        # Clear old gradients
        optimizer.zero_grad()

        # Forward pass
        outputs = model(images)

        # Calculate loss
        loss = criterion(outputs, labels)

        # Backpropagation
        loss.backward()

        # Update trainable weights
        optimizer.step()

        # Track loss and training accuracy
        running_loss += loss.item() * images.size(0)

        predictions = torch.argmax(outputs, dim=1)

        correct += (predictions == labels).sum().item()
        total += labels.size(0)

        progress_bar.set_postfix(loss=loss.item())

    epoch_loss = running_loss / total
    epoch_accuracy = 100 * correct / total

    print(f"\nEpoch {epoch+1}/{num_epochs}")
    print(f"Training Loss: {epoch_loss:.4f}")
    print(f"Training Accuracy: {epoch_accuracy:.2f}%")
```

```python
import os
import re
import urllib.request
from html.parser import HTMLParser

# Official NLM image directory
base_url = (
    "https://data.lhncbc.nlm.nih.gov/public/"
    "Tuberculosis-Chest-X-ray-Datasets/"
    "Montgomery-County-CXR-Set/"
    "MontgomerySet/CXR_png/"
)

save_dir = "/content/montgomery_dataset/CXR_png"
os.makedirs(save_dir, exist_ok=True)

class LinkParser(HTMLParser):
    def __init__(self):
        super().__init__()
        self.files = []

    def handle_starttag(self, tag, attrs):
        if tag.lower() == "a":
            for key, value in attrs:
                if key.lower() == "href" and value:
                    filename = value.split("/")[-1]

                    if re.fullmatch(
                        r"MCUCXR_\d+_[01]\.png",
                        filename,
                        flags=re.IGNORECASE
                    ):
                        self.files.append(filename)

try:
    request = urllib.request.Request(
        base_url + "index.html",
        headers={"User-Agent": "Mozilla/5.0"}
    )

    with urllib.request.urlopen(request, timeout=60) as response:
        html = response.read().decode("utf-8", errors="replace")

    parser = LinkParser()
    parser.feed(html)

    filenames = sorted(set(parser.files))

    print("Image files found:", len(filenames))

    if not filenames:
        print("No image links found in the directory listing.")

    else:
        downloaded = 0
        failed = []

        for i, filename in enumerate(filenames, start=1):
            destination = os.path.join(save_dir, filename)

            if os.path.exists(destination) and os.path.getsize(destination) > 0:
                downloaded += 1
                continue

            try:
                image_request = urllib.request.Request(
                    base_url + filename,
                    headers={"User-Agent": "Mozilla/5.0"}
                )

                with urllib.request.urlopen(
                    image_request, timeout=60
                ) as response:
                    data = response.read()

                if not data:
                    raise ValueError("Empty response")

                with open(destination, "wb") as file:
                    file.write(data)

                downloaded += 1

            except Exception as error:
                failed.append((filename, str(error)))

            if i % 10 == 0 or i == len(filenames):
                print(
                    f"Progress: {i}/{len(filenames)} | "
                    f"Downloaded/available: {downloaded} | "
                    f"Failed: {len(failed)}"
                )

        print("\nDownload process finished.")
        print("Downloaded or already available:", downloaded)
        print("Failed:", len(failed))

        if failed:
            print("First few failures:", failed[:5])

except Exception as error:
    print("Directory access failed:", repr(error))
```

```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from torchvision import transforms

# Image preprocessing
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# Dataset class
class CombinedChestXrayDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, index):
        image = Image.open(self.paths[index]).convert("RGB")
        label = int(self.labels[index])

        if self.transform:
            image = self.transform(image)

        return image, label

# Create datasets
train_combined_dataset = CombinedChestXrayDataset(
    train_paths, train_labels, train_transform
)

val_combined_dataset = CombinedChestXrayDataset(
    val_paths, val_labels, eval_transform
)

test_combined_dataset = CombinedChestXrayDataset(
    test_paths, test_labels, eval_transform
)

# Create data loaders
train_combined_loader = DataLoader(
    train_combined_dataset,
    batch_size=16,
    shuffle=True,
    num_workers=0
)

val_combined_loader = DataLoader(
    val_combined_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

test_combined_loader = DataLoader(
    test_combined_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

# Verify everything
print("Training batches:", len(train_combined_loader))
print("Validation batches:", len(val_combined_loader))
print("Testing batches:", len(test_combined_loader))

images, labels = next(iter(train_combined_loader))

print("\nImage batch shape:", images.shape)
print("Label batch shape:", labels.shape)
print("Labels in first batch:", labels.tolist())
```

```python
import torch
import torch.nn as nn
from torchvision import models

# Use the existing model if it is already loaded
# Otherwise, load the saved fine-tuned model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

if "model" not in globals():
    model = models.densenet121(weights=None)
    model.classifier = nn.Linear(model.classifier.in_features, 2)

    checkpoint_path = (
        "/content/drive/MyDrive/TB_Screening_Project/"
        "densenet121_tb_screening_finetuned.pth"
    )

    model.load_state_dict(
        torch.load(checkpoint_path, map_location="cpu", weights_only=True)
    )

# Freeze all feature layers
for param in model.features.parameters():
    param.requires_grad = False

# Unfreeze the final feature block
for param in model.features.denseblock4.parameters():
    param.requires_grad = True

for param in model.features.norm5.parameters():
    param.requires_grad = True

# Keep the classifier trainable
for param in model.classifier.parameters():
    param.requires_grad = True

model = model.to(device)

# Count trainable parameters
trainable_params = sum(
    p.numel() for p in model.parameters() if p.requires_grad
)

total_params = sum(p.numel() for p in model.parameters())

print("Device:", device)
print("Model: DenseNet121")
print("Output classes:", model.classifier.out_features)
print("Trainable parameters:", f"{trainable_params:,}")
print("Total parameters:", f"{total_params:,}")
print("Model preparation successful!")
```

```python
import copy
import torch
import torch.nn as nn
from torch.optim import Adam
from tqdm.auto import tqdm

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)

# Loss function
criterion = nn.CrossEntropyLoss()

# Different learning rates for feature layers and classifier
optimizer = Adam([
    {
        "params": model.features.denseblock4.parameters(),
        "lr": 1e-5
    },
    {
        "params": model.features.norm5.parameters(),
        "lr": 1e-5
    },
    {
        "params": model.classifier.parameters(),
        "lr": 1e-4
    }
])

# Training settings
NUM_EPOCHS = 10
best_val_loss = float("inf")
best_state = None

for epoch in range(NUM_EPOCHS):

    # -------------------------
    # TRAINING
    # -------------------------
    model.train()

    # Keep frozen feature layers in evaluation mode so their
    # batch-normalization statistics do not change.
    model.features.eval()
    model.features.denseblock4.train()
    model.features.norm5.eval()
    model.classifier.train()

    train_loss = 0.0
    train_correct = 0
    train_total = 0

    for images, labels in tqdm(
        train_combined_loader,
        desc=f"Epoch {epoch + 1}/{NUM_EPOCHS}"
    ):
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model(images)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        train_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)
        train_correct += (predictions == labels).sum().item()
        train_total += labels.size(0)

    train_loss /= train_total
    train_accuracy = 100 * train_correct / train_total

    # -------------------------
    # VALIDATION
    # -------------------------
    model.eval()

    val_loss = 0.0
    val_correct = 0
    val_total = 0

    with torch.no_grad():
        for images, labels in val_combined_loader:
            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)
            loss = criterion(outputs, labels)

            val_loss += loss.item() * images.size(0)

            predictions = outputs.argmax(dim=1)
            val_correct += (predictions == labels).sum().item()
            val_total += labels.size(0)

    val_loss /= val_total
    val_accuracy = 100 * val_correct / val_total

    print(f"\nEpoch {epoch + 1}/{NUM_EPOCHS}")
    print(f"Training loss:   {train_loss:.4f}")
    print(f"Training accuracy: {train_accuracy:.2f}%")
    print(f"Validation loss: {val_loss:.4f}")
    print(f"Validation accuracy: {val_accuracy:.2f}%")

    # Save the best model according to validation loss
    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())

        print("New best model found!")

# Restore the best validation checkpoint
if best_state is not None:
    model.load_state_dict(best_state)

# Save checkpoint to Google Drive
import os

save_dir = "/content/drive/MyDrive/TB_Screening_Project"
os.makedirs(save_dir, exist_ok=True)

best_model_path = os.path.join(
    save_dir,
    "densenet121_combined_best.pth"
)

torch.save(model.state_dict(), best_model_path)

print("\nTraining completed!")
print("Best validation loss:", round(best_val_loss, 4))
print("Saved model to:", best_model_path)
```

```python
import os
import requests
from concurrent.futures import ThreadPoolExecutor
from tqdm.auto import tqdm

# Dataset folders
shenzhen_dir = "/content/tb_dataset"
montgomery_dir = "/content/montgomery_dataset/CXR_png"

os.makedirs(shenzhen_dir, exist_ok=True)
os.makedirs(montgomery_dir, exist_ok=True)

datasets = [
    (
        "https://data.lhncbc.nlm.nih.gov/public/"
        "Tuberculosis-Chest-X-ray-Datasets/"
        "Shenzhen-Hospital-CXR-Set/CXR_png/",
        shenzhen_dir,
        662
    ),
    (
        "https://data.lhncbc.nlm.nih.gov/public/"
        "Tuberculosis-Chest-X-ray-Datasets/"
        "Montgomery-County-CXR-Set/MontgomerySet/CXR_png/",
        montgomery_dir,
        138
    )
]

def download_image(item):
    url, folder, filename = item
    path = os.path.join(folder, filename)

    if os.path.exists(path) and os.path.getsize(path) > 0:
        return True

    try:
        response = requests.get(url, timeout=30)
        response.raise_for_status()

        if not response.content.startswith(b"\x89PNG\r\n\x1a\n"):
            return False

        with open(path, "wb") as f:
            f.write(response.content)

        return True
    except Exception:
        return False

for base_url, folder, expected_count in datasets:
    dataset_name = os.path.basename(os.path.dirname(folder))
    print(f"\nRestoring images in: {folder}")

    # Read filenames from the official NLM directory index.
    index_url = base_url + "index.html"
    response = requests.get(index_url, timeout=30)
    response.raise_for_status()

    from bs4 import BeautifulSoup
    soup = BeautifulSoup(response.text, "html.parser")

    filenames = sorted({
        os.path.basename(link.get("href", "").split("?")[0])
        for link in soup.find_all("a", href=True)
        if link.get("href", "").lower().endswith(".png")
    })

    print("Image files listed:", len(filenames))

    if not filenames:
        print("No image filenames found. Stop and share this output.")
        continue

    tasks = [
        (base_url + filename, folder, filename)
        for filename in filenames
    ]

    with ThreadPoolExecutor(max_workers=8) as executor:
        results = list(tqdm(
            executor.map(download_image, tasks),
            total=len(tasks),
            desc=dataset_name
        ))

    successful = sum(results)
    print(f"Successfully available: {successful}/{len(filenames)}")

print("\nFinal folder counts:")
for folder in [shenzhen_dir, montgomery_dir]:
    count = sum(
        f.lower().endswith(".png")
        for f in os.listdir(folder)
    )
    print(folder, "->", count, "PNG files")
```

```python
import os
import torch
import torch.nn as nn
from PIL import Image
from torchvision import transforms, models
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import train_test_split
from collections import Counter

# 1. Dataset folders
shenzhen_dir = "/content/tb_dataset"
montgomery_dir = "/content/montgomery_dataset/CXR_png"

# 2. Collect image paths and labels
all_image_paths = []
all_labels = []

for folder in [shenzhen_dir, montgomery_dir]:
    for filename in sorted(os.listdir(folder)):
        if not filename.lower().endswith(".png"):
            continue

        if filename.endswith("_0.png"):
            label = 0
        elif filename.endswith("_1.png"):
            label = 1
        else:
            continue

        all_image_paths.append(os.path.join(folder, filename))
        all_labels.append(label)

print("Total images:", len(all_image_paths))
print("Class counts:", Counter(all_labels))

# 3. Recreate the same seeded split method
train_paths, remaining_paths, train_labels, remaining_labels = (
    train_test_split(
        all_image_paths,
        all_labels,
        test_size=0.30,
        random_state=42,
        stratify=all_labels
    )
)

val_paths, test_paths, val_labels, test_labels = (
    train_test_split(
        remaining_paths,
        remaining_labels,
        test_size=0.50,
        random_state=42,
        stratify=remaining_labels
    )
)

print("\nTraining:", len(train_paths))
print("Validation:", len(val_paths))
print("Testing:", len(test_paths))

# 4. Image preprocessing for evaluation
eval_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

# 5. Dataset class
class CombinedChestXrayDataset(Dataset):
    def __init__(self, paths, labels, transform=None):
        self.paths = paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.paths)

    def __getitem__(self, index):
        image = Image.open(self.paths[index]).convert("RGB")
        label = int(self.labels[index])

        if self.transform:
            image = self.transform(image)

        return image, label

# 6. Recreate test loader
test_combined_dataset = CombinedChestXrayDataset(
    test_paths,
    test_labels,
    eval_transform
)

test_combined_loader = DataLoader(
    test_combined_dataset,
    batch_size=16,
    shuffle=False,
    num_workers=0
)

print("\nTest loader ready!")
print("Test batches:", len(test_combined_loader))
```

```python
import os
import numpy as np
import pandas as pd
from PIL import Image
from tqdm.auto import tqdm

datasets = {
    "Shenzhen": "/content/tb_dataset",
    "Montgomery": "/content/montgomery_dataset/CXR_png"
}

results = []

for dataset_name, folder in datasets.items():
    files = sorted(
        f for f in os.listdir(folder)
        if f.lower().endswith(".png")
    )

    for filename in tqdm(files, desc=dataset_name):
        path = os.path.join(folder, filename)

        with Image.open(path) as img:
            image = np.asarray(
                img.convert("L").resize((224, 224)),
                dtype=np.uint8
            )

        # Fraction of pixels that are very dark
        dark_fraction = np.mean(image < 20)

        results.append({
            "dataset": dataset_name,
            "filename": filename,
            "dark_pixel_fraction": float(dark_fraction)
        })

df_background = pd.DataFrame(results)

print("Images analyzed:", len(df_background))
print("\nDark-background summary:")
print(
    df_background.groupby("dataset")["dark_pixel_fraction"]
    .agg(["mean", "median", "std"])
    .round(4)
)
```

```python
%%writefile /content/app.py

import streamlit as st
import torch
import torch.nn as nn
from torchvision import models, transforms
from PIL import Image

MODEL_PATH = (
    "/content/drive/MyDrive/TB_Screening_Project/"
    "densenet121_combined_best.pth"
)

st.set_page_config(
    page_title="TB Screening Assistant",
    page_icon="🩻",
    layout="wide"
)

st.title("🩻 TB Screening Assistant")
st.write("AI-assisted chest X-ray screening prototype")

st.warning(
    "Research prototype only. This tool cannot diagnose or rule out "
    "tuberculosis. Results require assessment by a qualified healthcare "
    "professional."
)


@st.cache_resource
def load_model():
    device = torch.device(
        "cuda" if torch.cuda.is_available() else "cpu"
    )

    model = models.densenet121(weights=None)
    model.classifier = nn.Linear(
        model.classifier.in_features, 2
    )

    state = torch.load(
        MODEL_PATH,
        map_location=device,
        weights_only=True
    )

    model.load_state_dict(state)
    model.to(device)
    model.eval()

    return model, device


preprocess = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

uploaded_file = st.file_uploader(
    "Upload a chest X-ray",
    type=["png", "jpg", "jpeg"]
)

if uploaded_file is not None:
    try:
        original_image = Image.open(uploaded_file).convert("RGB")

        # Display the original X-ray and its resized preview.
        processed_image = original_image.resize(
            (224, 224),
            Image.Resampling.BILINEAR
        )

        col1, col2 = st.columns(2)

        with col1:
            st.subheader("Original X-ray")
            st.image(original_image, use_container_width=True)

        with col2:
            st.subheader("Resized X-ray (224 × 224)")
            st.image(processed_image, use_container_width=True)

        if st.button("Run screening", type="primary"):
            with st.spinner("Processing X-ray..."):
                model, device = load_model()

                input_tensor = preprocess(
                    original_image
                ).unsqueeze(0).to(device)

                with torch.no_grad():
                    logits = model(input_tensor)
                    probabilities = torch.softmax(logits, dim=1)[0]

                predicted_class = int(
                    torch.argmax(probabilities).item()
                )
                confidence = float(
                    probabilities[predicted_class].item()
                )

            st.subheader("Experimental screening result")

            if predicted_class == 1:
                st.error("Abnormal-pattern classification")
            else:
                st.success("Normal-pattern classification")

            st.write(
                f"Model confidence for this predicted class: "
                f"{confidence * 100:.1f}%"
            )

            st.info(
                "This result is not a diagnosis. A normal-pattern "
                "classification does not rule out tuberculosis. "
                "Follow appropriate clinical assessment and testing "
                "guidance."
            )

    except Exception as error:
        st.error(f"Unable to process the image: {error}")

st.divider()

st.caption(
    "Demonstration prototype — not validated for independent "
    "clinical decision-making."
)
```
